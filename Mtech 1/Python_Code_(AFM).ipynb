Write a python script to train any classifier (KNN/MLP/SVM/RF) and predict test
samples with the trained model.

```python
# P1: Train a classifier and predict test samples

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
# You can import others:
# from sklearn.svm import SVC
# from sklearn.neural_network import MLPClassifier
# from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score

# Load dataset
data = load_iris()
X, y = data.data, data.target

# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Choose classifier (KNN example)
model = KNeighborsClassifier(n_neighbors=3)

# Train
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Accuracy
acc = accuracy_score(y_test, y_pred)

print("Predicted:", y_pred)
print("Actual:   ", y_test)
print("Accuracy: ", acc)
```

Consider a dataset from sklearn and apply K-Means clustering. Also compute
clustering accuracy.

```python
# P2: K-Means clustering and compute accuracy

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score
import numpy as np

# Load data
data = load_iris()
X, y = data.data, data.target

# Apply K-Means
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

# Cluster labels
labels = kmeans.labels_

# KMeans label numbers may not match true label numbers
# So we map cluster labels to actual labels
def cluster_accuracy(true, pred):
    mapping = {}
    for cluster in np.unique(pred):
        idx = pred == cluster
        majority_label = np.bincount(true[idx]).argmax()
        mapping[cluster] = majority_label
    mapped_preds = np.array([mapping[p] for p in pred])
    return accuracy_score(true, mapped_preds)

acc = cluster_accuracy(y, labels)

print("K-Means Cluster Labels:", labels)
print("Clustering Accuracy:", acc)
```

Consider the following data. Apply linear regression and find the value of Y when
X=7.7. Visualize with suitable plot.

```python
# P3: Linear Regression and prediction at X = 7.7

import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Given data
X = np.array([2.1, 3.5, 4.2, 4.7, 5.6, 5.8, 6.5]).reshape(-1, 1)
Y = np.array([4.7, 5.5, 6.4, 8.0, 9.1, 9.4, 9.2])

# Train linear regression
model = LinearRegression()
model.fit(X, Y)

# Predict Y when X = 7.7
x_new = np.array([[7.7]])
y_pred = model.predict(x_new)

print("Predicted Y for X = 7.7:", y_pred[0])

# Visualization
plt.scatter(X, Y, label="Data points")
plt.plot(X, model.predict(X), label="Regression Line")
plt.scatter([7.7], y_pred, color='red', label="Prediction (7.7, %.2f)" % y_pred)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.title("Linear Regression")
plt.show()
```
