```r
data(mtcars)

min_mpg <- min(mtcars$mpg)
max_mpg <- max(mtcars$mpg)
mean_mpg <- mean(mtcars$mpg)
median_mpg <- median(mtcars$mpg)
mode_mpg <- as.numeric(names(sort(table(mtcars$mpg), decreasing = TRUE)[1]))

cat("Min MPG:", min_mpg, "\n")
cat("Max MPG:", max_mpg, "\n")
cat("Mean MPG:", mean_mpg, "\n")
cat("Median MPG:", median_mpg, "\n")
cat("Mode MPG:", mode_mpg, "\n")

barplot(mtcars$mpg, main = "Distribution of MPG", xlab = "Cars", ylab = "MPG", col = "skyblue")
```

```r
# Install required packages (only run once)
install.packages("e1071")
install.packages("caret")

# Load libraries
library(e1071)
library(caret)

# Load data
data(iris)

# Split dataset
set.seed(123)
trainIndex <- createDataPartition(iris$Species, p = 0.7, list = FALSE)
trainData <- iris[trainIndex, ]
testData <- iris[-trainIndex, ]

# Build SVM model
svm_model <- svm(Species ~ ., data = trainData, kernel = "linear")

# Predict
predictions <- predict(svm_model, testData)

# Evaluate
confusionMatrix(predictions, testData$Species)
```
