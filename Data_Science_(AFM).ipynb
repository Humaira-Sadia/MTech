# Panda Operations

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv')

print(f"First 5 rows : {df.head()}")
print(f"Last 5 rows : {df.tail()}")
print(f"Shape of the data : {df.shape}")
print(f"Defination : {df.describe()}")
print(f"Info : {df.info()}")
```

# Data Visualization

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="ticks", color_codes=True)

tips = sns.load_dataset("tips")

sns.histplot(tips["total_bill"])
plt.show()

sns.histplot(tips["total_bill"], kde=True)
plt.show()

sns.boxplot(x="day", y="total_bill", data=tips)
plt.show()
```

# Statistics

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = np.random.randint(1,101,100)

print(f"Mean : {np.mean(data)}")
print(f"Median : {np.median(data)}")
print(f"Variance : {np.var(data)}")
print(f"Standard Deviation : {np.std(data)}")
print(f"Correlation : {np.corrcoef(data)}")

plt.hist(data)
plt.show()
```

# Pearson Correlation

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.pairplot(df, hue="species")
plt.show()

print(f"Correlation : {df.corr(numeric_only=True)}")
```

# Spearman Correlation

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.pairplot(df, hue="species")
plt.show()

spearman_corr = df.corr(method='spearman',numeric_only=True)
print(f"Correlation : {spearman_corr}")

sns.heatmap(spearman_corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```

# Linear Regression

```python
from sklearn.linear_model import LinearRegression

X = df['sepal_length']
y = df['petal_length']

model = LinearRegression()
model.fit(X.values.reshape(-1, 1), y.values.reshape(-1, 1))

print(f"Slope : {model.coef_}")
print(f"Intercept : {model.intercept_}")
```
